```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\Junayed\pandas_prac\Aug-1\messy_aug1.csv')
```

```python
print('Shape:',df.shape)
print('Data Type:')
df.dtypes
```

```python
df['Qty'] = df['Qty'].astype("Int64")
df['Qty']
```

```python
df['UnitPrice'] = df['UnitPrice'].astype(str).str.replace("$","",regex=False).str.strip()

df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')

df['UnitPrice']
```

```python
df['Discount'] = df['Discount'].astype(str).str.replace("%","",regex=False).str.strip()

df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce') / 100

df['Discount']
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], format='mixed')
```

```python
df['Status'] = df['Status'].astype(str).str.strip().str.title()

df['Status']
```

```python
df['Status'].value_counts()
```

```python
import numpy as np
df['Comment'] = df['Comment'].astype(str).str.strip()

df['Comment'] = df['Comment'].replace(['unknown','Missing','None','nan'],np.nan)
```

```python
df['Comment']
```

```python
sub_set = ["Qty","UnitPrice","Discount","OrderDate"]

dup_chk = df.duplicated(sub_set, keep=False)
df[dup_chk]
```

```python
df = df[~df['OrderID'].isin([5006,5013])]
```

```python
df['Customer'] = df['Customer'].fillna("Unknown")
df.reset_index(drop=True)
```

```python
df['Discount'] = df['Discount'].fillna(0)

df['Discount']
```

```python
df['Final Price'] = (df['Qty']*df['UnitPrice']) * (1- df['Discount'])
```

```python
target_col = df.columns.get_loc("Discount")
df.insert(target_col + 1, "Final Price", df.pop("Final Price"))
```
