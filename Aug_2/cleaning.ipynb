```python
df["Phone"] = df["Phone"].astype(str).str.strip()
```

```python
import re

def clean_phone_num(val):
    if pd.isna(val):
        return np.nan
    digits = re.sub(r"\D","", str(val))

    if len(digits) == 11 and digits.startswith("1"):
        digits = digits[1:]
    return digits if len(digits) == 10 else np.nan

df["Phone"] = df["Phone"].apply(clean_phone_num)
df["Phone"]
```

```python
df["Phone"] = df["Phone"].apply(lambda x: str(x))
df["Phone"] = df["Phone"].apply(lambda x: x[0:3] + "-" + x[3:6] + "-" + x[6:10])

df["Phone"]
```

```python
df["Phone"] = df["Phone"].astype(str).str.replace("nan--", "Unknown")

df["Phone"]
```

```python
df["Items"] = df["Items"].apply(
    lambda x: [item.strip().title() for item in str(x).split(",")]
)

df["Items"]
```

```python
df["OrderTotal"] = pd.to_numeric(df["OrderTotal"].astype(str).str.replace("$","", regex=False).str.strip(), errors='coerce')

df["OrderTotal"]
```

```python
df.loc[df["OrderTotal"] <0 , "OrderTotal"] = abs(df["OrderTotal"])

df["OrderTotal"]
```

```python
tip_map = {
    "yes": True, "y": True, "1": True, "true": True,
    "no": False, "n": False, "0": False, "false": False
}

df["TipGiven"] = df["TipGiven"].astype(str).str.strip().str.lower().map(tip_map)
```

```python
def clean_rating(val):
    val = str(val).strip().lower()
    if val in ("nan", "n/a", ""):
        return np.nan
    if "/" in val:
        num, denom = val.split("/")
        return round(float(num) / float(num) *5 ,1)

    try:
        match = "".join(char for char in str(val) if char.isdigit() or char == ".")
        return float(match)
    except ValueError:
        return np.nan

df["Rating"] = df["Rating"].apply(clean_rating)
df["Rating"]
```

```python
def clean_duration(val):
    if pd.isna(val):
        return np.nan

    val = str(val).strip().lower()
    h_match = re.search(r"(\d+)\s*h", val)
    m_match = re.search(r"(\d+)\s*m", val)
    if h_match or m_match:
        hours = int(h_match.group(1) if h_match else 0)
        minutes = int(m_match.group(1) if m_match else 0)
        return hours * 60 + minutes
    num_match = re.search(r"\d+", val)
    return int(num_match.group() if num_match else np.nan)

df["DeliveryTime"] = df["DeliveryTime"].apply(clean_duration)

df.loc[df["DeliveryTime"] > 180 , "DeliveryTime"] = np.nan

df["DeliveryTime"]
```

```python
pay_map = {
    "credit card": "Credit Card", "cc": "Credit Card",
    "debit card": "Debit Card",
    "cash": "Cash",
}

df["PaymentMethod"] = (df["PaymentMethod"].astype(str).str.strip().str.lower().str.replace("_", " ", regex=False).map(pay_map))

df["PaymentMethod"]
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"], format= 'mixed', dayfirst=True).dt.strftime("%Y-%m-%d")

df["OrderDate"]
```

```python
category_map = {
    "fast food": "Fast Food",
    "fastfood": "Fast Food",
    "fast-food": "Fast Food",
    "fine dining": "Fine Dining",
    "cafe": "Cafe"
}

df["Category"] = df["Category"].astype(str).str.strip().str.lower()
df["Category"] = df["Category"].map(category_map)
df["Category"]
```

```python
df["Zip"] = df["Address"].str.extract(r"(\d{5})$")
df[["Address", "Zip"]]
```
