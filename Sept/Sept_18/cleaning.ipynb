### Clean Member Names
- Converted column values to text (`astype(str)`)
- Removed leading and trailing whitespace (`strip()`)
- Formatted names into Title Case (`title()`)


```python
df["MemberName"] = df["MemberName"].astype(str).str.strip().str.title()
df["MemberName"]
```

```python
df['MemberEmail'].isna().sum()
```

### Validate Email Format
- Defined a **regex pattern** to check for valid email structures
- Used `.str.match()` to test each email against the pattern
- Filtered to find rows where the email **failed** the check (`~valid_email`) but is not missing


```python
email_pattern = r'^[\w.\-]+@[\w\-]+\.[a-zA-Z]{2,}$'
valid_email = df["MemberEmail"].str.match(email_pattern, na=False)
df.loc[~valid_email & df["MemberEmail"].notna(), ["BookingID", "MemberName", "MemberEmail"]]
```

### Update Emails for Specific Booking IDs
- Used `.isin()` to select rows matching a list of Booking IDs
- Assigned corresponding email addresses to the `MemberEmail` column


```python
df.loc[df["BookingID"].isin(["B004","B009","B029","B034"]), "MemberEmail"] = ["kwame.asante@mail.com","amara.chukwu@mail.com", "kwame.asante@mail.com","amara.chukwu@mail.com"]
```

### Clean Phone Numbers
- Converted values to strings and removed non-digit characters (`re.sub(r"\D", ...)`)
- Formatted 10-digit numbers into standard format (`XXX-XXX-XXXX`)
- Handled missing values gracefully (`pd.isna` and `np.nan`)


```python
def clean_phone(val):
    if pd.isna(val):
        return np.nan
    digits = re.sub(r"\D", "", str(val))
    if len(digits) == 10:
        return f"{digits[0:3]}-{digits[3:6]}-{digits[6:10]}"
    return np.nan

df["MemberPhone"] = df["MemberPhone"].apply(clean_phone)
df["MemberPhone"]
```
