### Validate Email Format
- Defined a **regex pattern** to check for valid email structures
- Used `.str.match()` to test each email against the pattern

```python
email_pattern = r'^[\w.\-]+@[\w\-]+\.[a-zA-Z]{2,}$'
valid_email = df["MemberEmail"].str.match(email_pattern, na=False)
df.loc[~valid_email & df["MemberEmail"].notna(), ["BookingID", "MemberName", "MemberEmail"]]
```

### Clean Phone Numbers
- Converted values to strings and removed non-digit characters (`re.sub(r"\D", ...)`)
- Formatted 10-digit numbers into standard format (`XXX-XXX-XXXX`)
- Handled missing values gracefully (`pd.isna` and `np.nan`)


```python
def clean_phone(val):
    if pd.isna(val):
        return np.nan
    digits = re.sub(r"\D", "", str(val))
    if len(digits) == 10:
        return f"{digits[0:3]}-{digits[3:6]}-{digits[6:10]}"
    return np.nan

df["MemberPhone"] = df["MemberPhone"].apply(clean_phone)
df["MemberPhone"]
```

a misspelling (`"Private Ofice"`).

**Fix Applied:** `.str.strip()` alone won't collapse an internal double space used
`.str.replace(r"\s+", " ", regex=True)` first to normalize internal whitespace, then the
usual title-case + typo map.

```python
df["DeskType"].value_counts()
```

```python
df["DeskType"] = df["DeskType"].astype(str).str.strip().str.title()
df["DeskType"].value_counts()
```

```python
df["DeskType"] = df["DeskType"].astype(str).str.strip().str.replace(r"\s+", " ", regex=True)
desk_fix = {"Private Ofice": "Private Office"}
df["DeskType"] = df["DeskType"].replace(desk_fix)
df["DeskType"].value_counts()
```

## `HoursBooked`: Mixed-Unit Parsing

**Issue Identified:** 
written as `"4 hrs"`, `"3h"`, a bare `"2.5"`, or `"480 min"`. One row has no value logged at
all.

**Fix Applied:** Detect the `"min"` suffix, divide by 60 when present, otherwise treat the
extracted number as already being in hours. The missing row becomes `NaN`.

```python
is_minutes = df["HoursBooked"].astype(str).str.contains("min", case=False, na=False)
raw_number = df["HoursBooked"].astype(str).str.extract(r"([\d.]+)")[0].astype(float)

df["HoursBooked"] = np.where(is_minutes, raw_number / 60, raw_number)
df["HoursBooked"]
```
