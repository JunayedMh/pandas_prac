```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df.loc[(df['Age'] < 10) | (df['Age'] > 100), 'Age'] = np.nan

df['Age']
```

```python
def height_to_cm(val):
    if pd.isna(val):
        return np.nan
    val = str(val).strip().lower()
    if val.endswith("cm"):
        return float(val.replace("cm",""))
    if val.endswith("ft"):
        num = float(val.replace("ft",""))
        feet = int(num)
        inches = round((num - feet ) * 10)
        return round(feet* 30.48 + inches * 2.54, 1)
    return np.nan

df["Height_cm"] = df["Height"].apply(height_to_cm)

df[['Height', 'Height_cm']]
```

```python
df['Height'] = df['Height_cm']
df = df.drop(columns=['Height_cm'])
```

```python
def weight_to_kg(val):
    if pd.isna(val):
        return np.nan

    val = str(val).strip().lower()

    if val.endswith("kg"):
        return float(val.replace("kg",""))

    if val.endswith("lbs"):
        return round(float(val.replace("lbs","")) * 0.453592, 1)
    return np.nan

df['Weight_kg'] = df['Weight'].apply(weight_to_kg)
df[['Weight', 'Weight_kg']]
```

```python
df['Weight'] = df['Weight_kg']

df = df.drop(columns = 'Weight_kg')
```

```python
df
```

```python
type_map = {
    "premium": "Premium", "premeium": "Premium",
    "basic": "Basic", "basc": "Basic",
    "vip": "VIP",
}
df["MembershipType"] = df["MembershipType"].str.strip().str.lower().map(type_map)
df["MembershipType"].value_counts()
```

```python
df["MonthlyFee"] = pd.to_numeric(df["MonthlyFee"].astype(str).str.replace("$","", regex=False).str.strip(), errors='coerce')

df["MonthlyFee"]
```

```python
df["JoinDate"] = pd.to_datetime(df["JoinDate"], format= 'mixed', dayfirst=False)

df["JoinDate"]
```

```python
df["Status"] = df["Status"].str.title()

df["Status"]
```

```python
df["EmailValid"] = df["Email"].astype(str).str.match(r"^[^@\s]+@[^@\s]+\.[^@\s]+$")

df['Emails'] = df['Email'].where(df['EmailValid'], np.nan)
df[["Email", "EmailValid", "Emails"]]
```

```python
df['Email'] = df['Emails']

df = df.drop(columns = 'Emails')
```

```python
df["Notes"] = df["Notes"].astype(str).str.strip().replace(["nan", "None", ""], np.nan)

df["Notes"]
```
