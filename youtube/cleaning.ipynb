```python
import pandas as pd
import numpy as np
import re
```

```python
drop_col = ["keyword", "channel_name", "n_videos", "is_verified", "keyword_matches", "matched_keywords"]
df = df.drop(drop_col, axis=1)
```

```python
df.head(5)
```

```python
def add_at(val):
    if pd.isna(val):
        return np.nan
    return f"@{val}"

df["handle"] = df["handle"].apply(add_at)
```

```python
df.head(5)
```

```python
engagement_sum = df["total_likes_last10"] + df["total_comments_last10"]
locking = df.columns.get_loc("avg_views_last10")
df.insert(locking + 1, "Total_engagement_sum", engagement_sum)
```

```python
df.head(5)
```

```python
drop_col1 = ["total_likes_last10", "total_comments_last10"]
df = df.drop(drop_col1, axis=1)
df.head(5)
```

```python
sorted_country = ["United States", "India", "Canada", "Indonesia", "Philippines"]

df = df[df["region"].isin(sorted_country)]
```
