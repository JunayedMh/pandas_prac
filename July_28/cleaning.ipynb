```python
df = df.drop_duplicates()
df
```

```python
df = df.drop("Not_Useful_Column",axis=1)
df
```

```python
df["Last_Name"]= df["Last_Name"].str.strip("123,./_ ")
df
```

```python
df["Names"] = (df["First_Name"] + " " + df["Last_Name"]).str.split().str.join(" ")
df
```

```python
full_name = df["Names"]
target_col = df.columns.get_loc("CustomerID")
df.insert(target_col+1,"Full_Name",full_name)
```

```python
df
```

```python
df = df.drop(columns=["Names","First_Name","Last_Name"])
df
```

```python
df["Phone_Number"] = df["Phone_Number"].astype(str).str.replace(r'[^0-9]', '', regex=True)
df
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: str(x))
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: x[0:3] + "-" + x[3:6] + "-" + x[6:10])
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace("nan--","")
df["Phone_Number"] = df["Phone_Number"].str.replace("--","")
```

```python
df
```

```python
df[["Street_Address", "State", "Zip_Code"]] = df["Address"].str.split(',',n=2, expand=True)
df
```

```python
street_name = df["Street_Address"]
state_name = df["State"]
zip_code = df["Zip_Code"]
target_col = df.columns.get_loc("Address")

df.insert(target_col + 1, "Street_Address", df.pop("Street_Address"))
df.insert(target_col + 2, "State", df.pop("State"))
df.insert(target_col + 3, "Zip_Code", df.pop("Zip_Code"))
df
```

```python
df = df.drop(columns= "Address")
df
```

```python
df["State"]= df["State"].fillna("").replace("NaN","")
df
```

```python
df["Zip_Code"]= df["Zip_Code"].fillna("").replace("NaN","")
df
```

```python
df["Paying Customer"] = df["Paying Customer"].str.replace("Yes","Y")
df["Paying Customer"] = df["Paying Customer"].str.replace("No","N")
df
```

```python
df["Street_Address"]= df["Street_Address"].str.replace("N/a","")
df["Paying Customer"]= df["Paying Customer"].str.replace("N/a","")
df
```

```python
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace("NaN","")
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace("No","N")
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace("Yes","Y")
df["Do_Not_Contact"] = df["Do_Not_Contact"].fillna("").replace("NaN","")
df
```

```python
for x in df.index:
    if df.loc[x,"Do_Not_Contact"] == "Y":
        df.drop(x,inplace=True)

df
```

```python
for x in df.index:
    if df.loc[x,"Phone_Number"] == "":
        df.drop(x,inplace=True)

df
```

```python
df = df.reset_index(drop=True)
df
```
