```python
df['SignupDate'] = pd.to_datetime(df['SignupDate'],format='mixed')
```

```python
df["SignupDate"] = pd.to_datetime(df["SignupDate"]).dt.strftime("%Y-%m-%d")
```

```python
df = df[df["Name"].notna()]
```

```python
df = df[df["Spend"].notna()]
```

```python
df
```

```python
active_map = {
    "Yes": "Yes", "yes": "Yes", "Y": "Yes", "y": "Yes", "1": "Yes", 1: "Yes",
    "No": "No", "no": "No", "N": "No", "n": "No", "0": "No", 0: "No"
}

df['Active'] = df['Active'].map(active_map)
```

```python
df
```

```python
df['Spend'] = df['Spend'].astype(str).str.replace("$","", regex=False).str.strip()
```

```python
df['Spend'] = pd.to_numeric(df['Spend'], errors = "coerce")
df['Spend'] = df['Spend'].map("${:,.2f}".format)
```

```python
df
```

```python
df['Notes'].value_counts()
```

```python
import numpy as np

df["Notes"] = df["Notes"].astype(str).str.strip()

df["Notes"] = df["Notes"].replace(
    ["unknown", "Missing", "missing", "nan", "NaN"], np.nan
)

df["Notes"] = df["Notes"].replace({"Regular customer": "Regular"})
```

```python
df["Notes"] = df["Notes"].fillna("Not Available")
```

```python
df
```

```python
df['Age'] = df['Age'].fillna("")
```

```python
df
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```
