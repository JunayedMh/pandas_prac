### Clean and Standardize Customer Names

* **Remove Extra Whitespace:** Strips leading and trailing spaces from text entries using `.str.strip()`.

### Clean and Format `CustomerPhone`

**Issues Identified:**
* Inconsistent phone number formatting across records (e.g., `(XXX) XXX-XXXX`, `XXX.XXX.XXXX`, `XXXXXXXXXX`, and `XXX-XXX-XXXX`).
* Presence of non-numeric punctuation, special characters, and missing values (`NaN`).

**Cleaning Steps:**
* **Missing Value Handling:** Returns `np.nan` immediately if the phone number entry is null (`pd.isna()`).
* **Digit Extraction:** Strips all non-numeric characters using `re.sub(r"\D", "", str(val))` to isolate raw digits.
* **Length Validation & Standardization:** Validates that the number contains exactly 10 digits and reformats it to `XXX-XXX-XXXX`. Any record that fails the length check is set to `np.nan`.
* **Application:** Applies the transformation to the `CustomerPhone` column and displays the series to verify changes.

```python
def Clean_PhoneNumber(val):
    if pd.isna(val):
        return np.nan
    digits = re.sub(r"\D", "", str(val))
    if len(digits) == 10:
        return f"{digits[0:3]}-{digits[3:6]}-{digits[6:10]}"
    return np.nan

df["CustomerPhone"] = df["CustomerPhone"].apply(Clean_PhoneNumber)
df["CustomerPhone"]
```

### Parse and Standardize `ServiceDate`

**Issues Identified:**
* Inconsistent date formats across entries, mixing ISO format (`YYYY-MM-DD`), US format (`MM/DD/YYYY`), and abbreviated text format (`DD-Mon-YYYY`).
* Stored as a generic text (`object`) data type rather than a native datetime object.

**Cleaning Steps:**
* **Flexible Date Parsing:** Uses `pd.to_datetime()` with `format='mixed'` to automatically infer and standardize varied date string patterns into a uniform `datetime64[ns]` type.
* **Error Handling:** Sets `errors="coerce"` to safely convert unparseable or corrupted date strings into `NaT` (Not a Time) instead of halting execution.
* **Verification:** Displays the converted `ServiceDate` column to confirm proper datetime formatting across all rows.

```python
df["ServiceDate"] = pd.to_datetime(df["ServiceDate"], format='mixed', errors="coerce")
df["ServiceDate"]
```
